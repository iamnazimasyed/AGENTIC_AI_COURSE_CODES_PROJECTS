```python
from transformers import AutoProcessor,AutoModel
processor = AutoProcessor.from_pretrained("suno/bark-small")
model = AutoModel.from_pretrained("suno/bark-small")
```

```python
inputs = processor(
    text = ["welcome to vampire diaries"],
    return_tensors='pt',

)
speech=model.generate(**inputs,do_sample=True)
```

```python
from IPython.display import Audio
sampling_rate=model.generation_config.sample_rate
Audio(speech.cpu().numpy().squeeze(),rate=sampling_rate)
```

```python
from transformers import AutoProcessor,MusicgenForConditionalGeneration
processor=AutoProcessor.from_pretrained("facebook/musicgen-small")
model=MusicgenForConditionalGeneration.from_pretrained("facebook/musicgen-small")

inputs=processor(
    text=["taylor swift lover album song"],
    padding=True,
    return_tensors="pt",


)
audio=model.generate(**inputs,max_new_tokens=1000)
```

```python
from IPython.display import Audio
sampling_rate=model.config.audio_encoder.sampling_rate
Audio(audio[0].numpy(),rate=sampling_rate)
```
