```python
import getpass
import os
from langchain_google_genai  import ChatGoogleGenerativeAI
#step1 creating Env for API KEY

if not os.environ.get("GOOGLE_API_KEY"):
    os.environ["GOOGLE_API_KEY"] = getpass.getpass("Enter your  API Key:")

#step2 Calling model
model=ChatGoogleGenerativeAI(model="gemini-2.5-flash")

#step3 Interaction with user and model
while True:
  user_input=input("You:")
  if user_input.lower()=="exit":
    print("Goodbye")
    break
  response=model.invoke(user_input)
  print("\nAi:",response.content)
```

```python
import getpass
import gradio as gr
import os
from langchain_google_genai  import ChatGoogleGenerativeAI
#step1 creating Env for API KEY

if not os.environ.get("GOOGLE_API_KEY"):
    os.environ["GOOGLE_API_KEY"] = getpass.getpass("Enter your  API Key:")

#step2 Calling model
model=ChatGoogleGenerativeAI(model="gemini-2.5-flash")

#steo 3 chat function
def chat_with_gemini(message,history):
    response = model.invoke(message)
    return response.content

#step 4 building interface using gradio
demo=gr.ChatInterface(
    fn=chat_with_gemini,
    title="Simple chat bot using gemini",
    description="Ask anything you want",
    theme="soft",


)
demo.launch(share=True)
```
