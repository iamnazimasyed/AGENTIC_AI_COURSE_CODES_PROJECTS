```python
# @title
from transformers import pipeline
import torch
from diffusers import DiffusionPipeline
generator=DiffusionPipeline.from_pretrained("CompVis/stable-diffusion-v1-4")
generator.to("cuda")
```

```python
prompt ="himalayas"
image = generator(prompt).images[0]
```
